### Drawing Molecules from SMILES using RDKit
We can use the `rdkit.Chem` and `rdkit.Chem.Draw` modules to parse SMILES strings and generate 2D molecular visualizations directly in the notebook.

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import IPythonConsole

# Enable inline RDKit visualization in Colab notebooks
IPythonConsole.ipython_useSVG = True

def draw_smiles(smiles_str, legend=""):
    """
    Parses a SMILES string and displays its standard 2D molecular visualization without explicit hydrogens.
    """
    mol = Chem.MolFromSmiles(smiles_str)
    if mol is None:
        print(f"Error: Invalid SMILES string: '{smiles_str}'")
        return None

    # Display the molecule without adding explicit hydrogens
    return Draw.MolsToGridImage([mol], molsPerRow=1, subImgSize=(350, 350), legends=[legend] if legend else None)
```

```python
# Example 2: Aspirin
aspirin_smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"
draw_smiles(aspirin_smiles, legend="Aspirin")
```

```python
ethylene_smiles = "C=C"
draw_smiles(ethylene_smiles, legend="Ethylene")
```

```python
propylene_smiles = "C=CC"
draw_smiles(propylene_smiles, legend="Propylene")
```

```python
from IPython.display import display

smiles_list = ["C=CC", "CC=CC", "CC(C)=CC", "c1ccccc1C=C", "c1cc(C=C)ccc1C=C"]
for smiles in smiles_list:
    img = draw_smiles(smiles, legend=smiles)
    display(img)
```
