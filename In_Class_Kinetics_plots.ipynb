## Theory of Chemical Kinetics

We will consider cases of zero order, first order and second order kinetics. In zero order kinetics, the reaction rate is independent of the concentration of the reactants. In first order kinetics, the reaction rate doubles when the concentration of the reactant doubles. In second order kinetics, the reaction rate increases 4-fold (quadruples) when the concentration of the reactant is doubled.

Here are the equations for the different reaction orders.

## Zero order

$Rate = k[A]^o = k$

The integrated rate equation is

$[A]_t = [A]_o - kt$

where

$[A]_o$ = initial concentration of A

$[A]_t$ = concentration of A at time t

k = the rate constant

## First order

$[A]_t = [A]_oe^{-kt}$

The integrated rate equation is

$ln\frac{[A]_t}{[A]_o} = -kt$

## Second order

$rate = k[A]^2$

The integrated rate equation is

$\frac{1}{[A]_t}-\frac{1}{[A]_o} = kt$
