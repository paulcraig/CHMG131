In programming, one of the first functions you learn to use is the ```print()``` statement. But what exactly does this function do?

In Python (and most other coding languages), the ```print()``` statement displays an input or value that you type in. This value is referred to as the **argument**. Your **argument** can be a few different values.

* Floating Point: a number with fractional parts (i.e. a decimal) [ex. 3.1415]
* Integer: Whole numbers that can be positive, negative, or 0. NEVER FRACTIONAL [ex. 1, -40, or 0]
* String: Letters or words typed in quotations [ex. "Hello World!"]

These three types of values are **data types** built into Python. Because they're built in, Python already knows what to recognize these types of values as; You don't need to indicate whether your value is a floating point, integer, or string.

Here we have an example of some values that need to be shown to the user. In order to display these values, we need to use a ```print()``` statement.

| My Pay | My Expenses |
|:------:|:----:|
|\$1000 | \$650 |
|\$1150 | \$750 |
|\$1050 | \$700 |

````
#If I want to display the first value in the column "My Pay", I can use this line of code
print(1000)
#But, if I want to print it with the dollar sign, rather than typing in 1000 (an integer) I would type "$1000", turning this into a string. This is because there is a dollar sign the computer's parser needs to interpret as a character.
print("$1000")
````
Now let's try to run these lines of code in the next code cell. Below, you will see these two statements already typed for you. All you have to do is press the play button at the top left of the cell.

``````python
print(1000)
print("$1000")
``````

Now let's try printing $1000, but without the quotation marks around it. What happens then?

``````python
print($1000)
``````

In the previous cell, the computer isn't able to understand the code typed in, because it's trying to read the **argument** as an integer. But since there's a dollar sign ($), the value isn't legible to the computer as an integer. So instead we would format the argument as a string as opposed to an integer.

Now let's get into variables. A **variable** is a label that corresponds to data. So what does that mean? Let's say I have the value $1000, but I want to store it as a name, let's say income. How would we go about this?

```
income = 1000
````

So in order to assign a **value** to a **variable**, you must write the name first, then an equal sign, and finally the value. The value can be an integer, float, or string.
Now, if you wanted to set a variable to $1000 (one thousand dollars with the dollar sign in front), you would have to format this as a string. If you recall from the previous section, the computer will interpret the dollar sign and the 1000 in this context as a string of characters, as opposed to an integer. This is because $ is not a numeric value. Here's how you would format setting a variable income to equal $1000

````
income = "$1000"
`````
